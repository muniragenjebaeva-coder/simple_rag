```python
# data
documents = [
    "DOMO kommunal toʻlovlar uchun mobil ilova.",
    "DOMO ilovasida gaz va elektr uchun toʻlash mumkin.",
    "DOMO orqali suv uchun toʻlash mumkin.",
    "DOMO ilovasini Google Play orqali yuklab olish mumkin.",
    "DOMO ofisi Toshkent shahrida joylashgan.",
]
documents
```

```python
# форматирование токенизация данных
def get_words(text):
    text =text.lower()
    text = text.replace(".", "")
    words = text.split()
    return words

print("форматирование токенизация данных:")
for doc in documents:
  print(get_words(doc))
```

```python
# оценка похожести
def score(question, doc):
  question_words = get_words(question)
  doc_words = get_words(doc)
  question_set = set(question_words)
  doc_set = set(doc_words)
  common = question_set & doc_set
  print(common)
  return len(common)
print("оценка похожести:")
print(score("gaz toʻlash", "DOMO ilovasida gaz va elektr uchun toʻlash mumkin."))
```

```python
# самый лучший ответ
def search(question):
  best_doc = None
  best_score = 0

  for doc in documents:
    score_for_doc = score(question, doc)
    if score_for_doc > best_score:
      best_score = score_for_doc
      best_doc = doc
  return best_doc
print("самый лучший ответ:")
print("gaz uchun qanday toʻlash mumkin: ", search("gaz uchun qanday toʻlash mumkin"))
print("ofis qayerda: ", search("ofis qayerda"))
```

```python
# суффиксный поиск
def similar(w1, w2):
  if len(w1) < 3 or len(w2) < 3:
    return False
  elif w1.startswith(w2) or w2.startswith(w1):
    return True
  return False
print("суффиксный поиск:")
print(similar("ofis", "ofisi"))
print(similar("ilovasida", "ilova"))
print(similar("gaz", "suv"))
print(similar("va", "vaqt"))
```

```python
def score(question, doc):
  question_words = set(get_words(question))
  doc_words = set(get_words(doc))
  count = 0
  for q in question_words:
    for w in doc_words:
      if similar(q, w):
        count += 1
        break
  return count
print(search("ofis qayerda"))
print(search("suvga"))
```

```python
q = "DOMO orqali gaz"
for doc in documents:
    print(score(q, doc), doc)
```

```python
# stop-words, частоповтооряющиеся слова
stop_words = {"domo", "bilan", "mumkin", "orqali", "uchun", "ham", "qayerda","qanday"}
def get_words(text):
  text = text.lower()
  text = text.replace(".", "")
  text = text.split()

  result = []
  for w in text:
    if w not in stop_words:
      result.append(w)
  return result
q = "DOMO orqali gaz"
for doc in documents:
    print(score(q, doc), doc)
```

```python
# честный ответ
def answer(question):
  doc = search(question)
  if doc == None:
    return "Javob topilmadi!"
  return doc
print(answer("suvga"))
print(answer("ofis qayerda"))
```
